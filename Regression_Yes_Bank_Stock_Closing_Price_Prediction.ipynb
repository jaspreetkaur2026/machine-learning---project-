### 8. Data Splitting

````python
# Chronological 80:20 train-test split
# Earlier observations are used for training and later observations are used for testing.
# This is more appropriate for time-dependent stock-price data.

split_index = int(len(X) * 0.80)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]
print("Training data shape:", X_train.shape, y_train.shape)
print("Testing data shape:", X_test.shape, y_test.shape)
````

##### What data splitting ratio have you used and why?

An 80:20 ratio was used to split the dataset. The first 80% of chronological observations were used for training, while the remaining 20% were kept as unseen test observations. This approach is appropriate for stock-price data because it preserves the time order and avoids using later observations to train a model that is evaluated on earlier observations.

#### 3. Explain each evaluation metric's indication towards business and the business impact pf the ML model used.



The four evaluation metrics considered for this regression project are **Mean Absolute Error (MAE), Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and R² Score**. These metrics help evaluate how accurately the model predicts Yes Bank's closing stock price.

**1. Mean Absolute Error (MAE):**
MAE represents the average absolute difference between the actual and predicted closing prices. A lower MAE indicates that the model's predictions are closer to the actual closing prices. From a business perspective, MAE is easy to understand because it represents the average prediction error in the same unit as the stock price. A lower MAE can support more accurate financial analysis and reduce the average difference between predicted and actual prices.

**2. Mean Squared Error (MSE):**
MSE measures the average squared difference between the actual and predicted closing prices. Because the errors are squared, larger prediction errors receive greater importance. Therefore, a lower MSE indicates better model performance and helps identify models that may produce large prediction errors. From a business perspective, minimizing MSE can be useful when larger prediction mistakes need to be given greater attention.

**3. Root Mean Squared Error (RMSE):**
RMSE is the square root of MSE and is expressed in the same unit as the stock price. It indicates the typical magnitude of prediction errors while giving greater importance to larger errors. A lower RMSE means that the model's predictions are generally closer to the actual closing prices. This makes RMSE useful for assessing the practical accuracy of the model and understanding the potential size of prediction errors.

**4. R² Score:**
R² Score indicates how effectively the model explains the variation in the closing price. A higher R² Score indicates that the model captures the relationship between the selected input variables and the closing price more effectively. A high R² can therefore indicate that the model is able to explain a large proportion of the observed variation in the target variable.


### Business Impact
The ML model provides a **data-driven framework for predicting Yes Bank's closing stock price** using the selected stock-price variables. Better prediction accuracy can support financial analysis, risk assessment, and data-driven decision-making. Lower MAE and RMSE indicate smaller prediction errors, while a lower MSE indicates fewer or less severe large errors. A higher R² indicates that the model explains more of the variation in the closing price.

These insights can help analysts better understand historical price behaviour and assess the predictive potential of Machine Learning models. However, stock markets are inherently uncertain, so the model should be considered a decision-support tool rather than a guarantee of future stock-price movements.

The general equation of the model is:

**Close = β₀ + β₁(Open) + β₂(High) + β₃(Low) + ε**

* **β₀** = intercept
* **β₁, β₂, β₃** = coefficients of the respective features
* **ε** = error term

Linear Regression was chosen because it is simple, computationally efficient, and highly interpretable. The coefficients make it possible to understand how each input feature influences the predicted closing price.
